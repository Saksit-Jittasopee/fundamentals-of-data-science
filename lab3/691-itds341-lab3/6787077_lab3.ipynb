```python
import numpy as np
import pandas as pd
```

```python
# Task 1: Read the titanic dataset
data_df = pd.read_csv('titanic-data-miss.csv')

# Check the number of NaN values in each column
print(data_df.isna().sum())
```

```python
# Task 2: Handle Missing Age
# Making a copy of our dataframe using copy
df2 = data_df.copy()
```

```python
# Task 2.1: Fill in age by a zero value
df2['Age'] = df2['Age'].fillna(0)

print(df2.isna().sum())
print(df2['Age'].mean())
```

```python
# Task 2.2: Fill in age by the mean
df2_2 = data_df.copy()

df2_2['Age'] = df2_2['Age'].fillna(df2_2['Age'].mean())

print(df2_2.isna().sum())
print(df2_2['Age'].mean())
```

```python
# Task 2.3: Fill in age by the IterativeImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imp = IterativeImputer(max_iter=10, random_state=0)
df2_3 = data_df.copy()

numeric_cols = df2_3.select_dtypes(include=['number']).columns

imputed_data = imp.fit_transform(df2_3[numeric_cols])

df2_3[numeric_cols] = imputed_data

print(df2_3.isna().sum())
print(df2_3['Age'].mean())
```

```python
# Task 2.4: Which one is the best method to handle the missing age data? and why?

## IterativeImputer is the best method among the three because it uses the relationships between 'Age' and other variables to make accurate, educated guesses, preserving the integrity of your dataset for machine learning models.
```

```python
# Task 3: Handle Missing Embarked

# Task 3.1: Remove passengers (i.e., rows) whose 'Embarked' are unknown
df3 = data_df.copy()
df3 = df3.dropna(subset=['Embarked'])

print(df3.isnull().sum())
print(len(df3))
```

```python
# Task 3.2: Fill in the missing Embarked with the mode value
df3_2 = data_df.copy()

embarked_mode = df3_2['Embarked'].mode()[0]

df3_2['Embarked'] = df3_2['Embarked'].fillna(embarked_mode)

print(df3_2.isnull().sum())
print(len(df3_2))
```

```python
# Task 3.3: Which one is the best method to handle the missing embarked data? and why?

## Remove the mssing embarked data is better method because the missing data in embarked only have 2 rows. It doesn't affected the dataset for machine learning model.
```

```python
# Task 4: Handle Missing Cabin
# Task 4.1: Fill in the missing Cabin with the mode value
df4 = data_df.copy()

cabin_mode = df4['Cabin'].mode()[0]

df4['Cabin'] = df4['Cabin'].fillna(cabin_mode)

print(df4.isnull().sum())
print(cabin_mode)
print(df4['Cabin'].value_counts())
```

```python
# Task 4.2: Fill in the missing cabin with 'UNK’
df4_2 = data_df.copy()

df4_2['Cabin'] = df4_2['Cabin'].fillna('UNK')

print(df4_2.isnull().sum())
print(df4_2['Cabin'].value_counts())
```

```python
# Task 4.3: Which one is the best method to handle the missing cabin data? and why?

## Fill in the missing cabin with 'UNK’ is a better method. A missing cabin number is not a random accident. The vast majority of passengers classes simply did not have cabin numbers recorded on their tickets, or the records were lost. Therefore, the absence of a cabin is a valuable piece of information that heavily correlates with passenger class and survival rate.
```

```python
# Task 5: Use the best approach from Task 2 to Task 4 to clean Age, Embarked, Cabin
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df5 = data_df.copy()

imp = IterativeImputer(max_iter=10, random_state=0)
numeric_cols = df5.select_dtypes(include=['number']).columns
df5[numeric_cols] = imp.fit_transform(df5[numeric_cols])

embarked_mode = df5['Embarked'].mode()[0]
df5['Embarked'] = df5['Embarked'].fillna(embarked_mode)

df5['Cabin'] = df5['Cabin'].fillna('UNK')

print(df5[['Age', 'Embarked', 'Cabin']].isna().sum())
print()
print(df5.isnull().sum())
```

```python
# Task 6: Handle Missing Sex
import pandas as pd
import numpy as np

clean_df = data_df.copy()

name2sex = {
    'Mr.' : 'male',
    'Mrs.': 'female',
    'Miss.' : 'female',
    'Mme': 'female',
    'Dr.': 'male',
    'Master': 'male',
}

def extract_sex(name):
    for title, sex_value in name2sex.items():
        if title in str(name):
            return sex_value
    return np.nan

clean_df['Sex'] = clean_df['Sex'].fillna(clean_df['Name'].apply(extract_sex))

print(clean_df.loc[data_df['Sex'].isna(), 'Sex'])
print(clean_df.isna().sum())
print(clean_df['Sex'].value_counts())
```

```python
# Task 7: Replace Pclass
pclass_name = {
    1: 'first',
		2: 'second',
		3: 'third'
}

clean_df['Pclass'] = clean_df['Pclass'].map(pclass_name)

print(clean_df['Pclass'].value_counts())
```

```python
# Task 8: Binning the age to create Age_group
import pandas as pd
import numpy as np

bins = [17, 25, 60, 100]

labels = ['Youth', 'Adult', 'Elderly']

clean_df['Age_group'] = pd.cut(clean_df['Age'], bins=bins, labels=labels, right=True)

print(clean_df['Age_group'].value_counts())
```
