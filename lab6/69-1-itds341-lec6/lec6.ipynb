```python
import numpy as np
import pandas as pd
```

```python
tips = pd.read_csv("tips.csv")

print(tips)
```

```python
tips.info()
```

```python
tips["tip_pct"] = tips["tip"] / tips["total_bill"]

print(tips[["total_bill", "tip", "tip_pct"]])
```

```python
print(tips.groupby("day")["tip_pct"].mean())
```

```python
tips['day'].unique()
```

```python
tips.groupby('day').size()
```

```python
by_day = tips.groupby("day")

print(type(by_day))
```

```python
print(by_day.get_group("Sun"))
```

```python
tips.groupby(["day", "smoker", "time"])["tip_pct"].mean()
```

```python
tips.groupby(["day", "smoker", "time"],as_index=False)[["tip_pct", "total_bill"]].mean()
```

```python
tips
```

```python
tips.groupby("day")["tip_pct"].agg(["count", "mean", "median", "std"])
```

```python
tips.groupby("day", as_index=False).agg(
        parties=("total_bill", "size"),
        average_bill=("total_bill", "mean"),
        median_tip_pct=("tip_pct", "median"),
        largest_party=("size", "max")
    )
```

```python
tips.groupby("day").agg({
    "tip": ["max", ("med","median")],         
    "total_bill": "mean" 
})
```

```python
def peak_to_peak(values):
    return values.max() - values.min()

tips.groupby("day")["total_bill"].agg(
    bill_range=peak_to_peak
)
```

```python
tips.groupby("day").agg({
    "tip": ["max", ("med","median")],         
    "total_bill": peak_to_peak
})
```

```python
tips.groupby("day")["total_bill"].agg(
    ["mean", "max", peak_to_peak]
)
```

```python
tips.groupby(
    ["time", "smoker"],
    as_index=False
).agg(
    parties=("size", "size"),
    average_bill=("total_bill", "mean"),
    median_tip_pct=("tip_pct", "median"),
    maximum_size=("size", "max")
)
```

```python
tips['day_mean_tip_pct'] = (
    tips.groupby("day")["tip_pct"]
    .transform("mean")
)
```

```python
tips
```

```python
(
    tips.groupby("day")["total_bill"]
    .transform(lambda x: (x - x.mean()) / x.std())
)
```

```python
tips.groupby(
    "day"
).filter(lambda group: len(group) >= 20)
```

```python
def top_bills(group, n=3):
    return group.nlargest(n, "total_bill")

(
    tips.groupby("day", group_keys=True) 
    .apply(top_bills, include_groups=False)
)
```

```python
tip_pivot = pd.pivot_table(
    tips,
    values="tip_pct",
    index="day",
    columns="smoker",
    aggfunc="mean"
)

tip_pivot
```

```python
renamed_pivot = tip_pivot.rename(columns={
    "No": "non_smoker",
    "Yes": "smoker"
})

renamed_pivot
```

```python
mixed_agg_pivot = pd.pivot_table(
    tips,
    values=["total_bill", "tip"],
    index="day",
    aggfunc={"total_bill": ["mean", "max"], "tip": "median"}
)
mixed_agg_pivot
```

```python
smoker_counts = pd.crosstab(
    tips["day"],
    tips["smoker"],
    margins=True
)
smoker_counts
```

```python
day_lookup = pd.DataFrame({
    "day": ["Thur", "Fri", "Sat", "Sun"],
    "period": ["weekday", "weekday", "weekend", "weekend"]
})

tips_with_period = tips.merge(
    day_lookup,
    on="day",
    how="left",
    validate="many_to_one"
)

print(tips_with_period[["day", "total_bill", "tip", "period"]])
```

```python
visits = pd.DataFrame({
    "day": pd.Categorical(
        ["Mon", "Mon", "Tue"],
        categories=["Mon", "Tue", "Wed"]
    )
})

print(visits.groupby("day", observed=True).size())
```
