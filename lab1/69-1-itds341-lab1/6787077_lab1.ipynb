```python
# Task 1: Create a DataFrame
data = {"total_bill": [16.99, 10.34, 21.01, 23.68, 24.59],
        "tip": [1.01, np.nan, 3.5, 3.31, 3.61],
        "smoker": ["No", "No", "Yes", np.nan, "No"], 
        "day": ["Sun", None, "Mon", "Wed", "Sat"],
        "time": ["Dinner", "Dinner", None, "Dinner", "Lunch"],
        "size": [2, 3, 3, 2, np.nan],
       }
df = pd.DataFrame(data)

df
```

```python
# T4: Use pd.read_csv to read the local tips.csv file into a DataFrame named df.
df = pd.read_csv('tips.csv')
df
```

```python
# T5-1: Select tipping data on Friday. Store your answer in result5_1.
result5_1 = df.loc[df['day'] == 'Fri']
print(result5_1)
```

```python
# T5-2: Select tipping data on the weekend (i.e., Saturday and Sunday).
result5_2 = df.loc[(df['day'] == 'Sat') | (df['day'] == 'Sun')]
print(result5_2)
```

```python
# T5-3: From T5-2, show only the 'total_bill' column.
result5_3 = df['total_bill']
print(result5_3)
```

```python
# T5-4: Select tipping data from the tables with a smoker (i.e., df['smoker']=='Yes') and the total bill is less than 10 (i.e., df['total_bill'] < 10).
result5_4 = df.loc[(df['smoker'] == 'Yes') & (df['total_bill'] < 10)]
print(result5_4)
```

```python
# T5-5: From T5-4: show only the 'tip','size' columns.
result5_5 = df[['tip', 'size']]
print(result5_5)
```

```python
# T6-1: Calculate the mean of the 'total_bill' during the weekend. Store your answer in result6_1.
df_weekend = df.loc[(df['day'] == 'Sat') | (df['day'] == 'Sun')]
result6_1 = df_weekend['total_bill'].mean()
print(result6_1)
```

```python
# T6-2: Calculate the mean of the 'tip' from the tables with a smoker.df_weekend = df.loc[(df['day'] > 'Sat') & (df['day'] == 'Sun')]
df_smoker = df.loc[(df['smoker'] == 'Yes')]
result6_2 = df_smoker['tip'].mean()
print(result6_2)
```

```python
# T6-3: Create a new column, named 'price_per_head', representing the price per head. This can be calculated by dividing the 'total_bill' with the 'size'.
df['price_per_head'] = df['total_bill'] / df['size']
df
```

```python
# T6-4: Calculate the maximum 'price_per_head' on Sunday.
result6_4 = df['price_per_head'].max()
print(result6_4)
```

```python
# T7-1: Sort by 'tip' in an ascending order. Store your answer in result7_1.
result7_1 = df.sort_values(by='tip', ascending=True)
print(result7_1)
```

```python
# T7-2: Sort by 'tip' in descending order.
result7_2 = df.sort_values(by='tip', ascending=False)
print(result7_2)
```

```python
# T7-3: Sort by 'total_bill' and 'tip' in descending order.
result7_3 = df.sort_values(by=['total_bill', 'tip'], ascending=False)
print(result7_3)
```

```python
# T7-4: Show the 'total_bill', 'tip' and 'price_per_head' columns during the weekend sorted by 'price_per_head' in an ascending order.
df_columns = df[['total_bill', 'tip', 'price_per_head']]
df_weekend = df_columns.loc[(df['day'] == 'Sat') | (df['day'] == 'Sun')]
result7_4 = df_weekend.sort_values(by=['price_per_head'], ascending=True)
print(result7_4)
```
