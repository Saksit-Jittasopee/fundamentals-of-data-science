```python
import pandas as pd
import numpy as np
```

```python
# Task 1: Create a DataFrame
data = {"total_bill": [16.99, 10.34, 21.01, 23.68, 24.59],
        "tip": [1.01, np.nan, 3.5, 3.31, 3.61],
        "smoker": ["No", "No", "Yes", np.nan, "No"], 
        "day": ["Sun", None, "Mon", "Wed", "Sat"],
        "time": ["Dinner", "Dinner", np.nan, "Dinner", "Lunch"],
        "size": [2, 3, 3, 2, np.nan],
       }
df = pd.DataFrame(data)

df
```

```python
# T2-1: Select 'tip' column using df[column].
result2_1 = df['tip']
result2_1
```

```python
# T2-2: Select 'size','tip','day' columns using df[column].
result2_2 = df[['size', 'tip', 'day']]
result2_2
```
