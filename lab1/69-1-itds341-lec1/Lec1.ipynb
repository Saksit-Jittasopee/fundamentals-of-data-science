```python
df4 = pd.DataFrame(np.random.standard_normal((4, 3)),
                  columns=list("bde"),
                  index=["Utah", "Ohio", "Texas", "Oregon"])
print(df4)
```

```python
def f1(x):
    return x.max() - x.min()

print(df4.apply(f1))
```

```python
print(df4.apply(f1, axis="columns"))
```

```python
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])

print(df4.apply(f2))
```
